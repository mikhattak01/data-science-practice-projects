Here’s a simple explanation of each **scikit-learn** function and what it measures — aimed at a data science beginner.

---

## 1. `accuracy_score`
**What it does:**  
Measures **how often** the model is correct.

**Formula:**  
(Number of correct predictions) / (Total predictions)

**When to use:**  
When classes are **balanced** (e.g., 50% spam, 50% not spam).

**Example:**  
If 90 out of 100 emails are classified correctly → accuracy = 0.90

---

## 2. `confusion_matrix`
**What it does:**  
Shows **where the model gets confused** — a table comparing actual vs predicted labels.

For **binary classification** (Yes/No):

|              | Predicted: No | Predicted: Yes |
|--------------|---------------|----------------|
| Actual: No   | TN            | FP              |
| Actual: Yes  | FN            | TP              |

- **TN** = True Negative (correctly said “No”)  
- **FP** = False Positive (wrongly said “Yes”)  
- **FN** = False Negative (wrongly said “No”)  
- **TP** = True Positive (correctly said “Yes”)

**Example:**  
```
[[50, 10],
 [ 5, 35]]
```
→ Model made 50 correct “No” predictions, 35 correct “Yes” predictions; got 10+5 wrong.

---

## 3. `precision_score`
**What it does:**  
Of all **positive predictions**, how many were **actually positive**?  
→ *“When model says Yes, how often is it right?”*

**Formula:**  
TP / (TP + FP)

**When to use:**  
When **false positives are costly** (e.g., spam filter marking good email as spam).

**Example:**  
Model predicted “Yes” 20 times, 15 were correct → precision = 15/20 = 0.75

---

## 4. `recall_score`
**What it does:**  
Of all **actual positives**, how many did the model **catch**?  
→ *“How sensitive is the model to finding Yes?”*

**Formula:**  
TP / (TP + FN)

**When to use:**  
When **false negatives are costly** (e.g., cancer detection – missing a sick person is terrible).

**Example:**  
There are 30 actual “Yes” cases, model found 20 → recall = 20/30 ≈ 0.67

---

## 5. `f1_score`
**What it does:**  
**Harmonic mean** of precision and recall.  
→ Balances both.

**Formula:**  
2 × (Precision × Recall) / (Precision + Recall)

**When to use:**  
When you need **one number** that balances precision & recall (especially with **imbalanced classes**).

**Example:**  
Precision=0.8, Recall=0.6 → F1 = 2×(0.8×0.6)/(0.8+0.6) = 0.96/1.4 ≈ 0.686

---

## 6. `classification_report`
**What it does:**  
Prints **precision, recall, f1-score, and support** (number of true instances) for each class in one table.

**Example output:**
```
              precision    recall  f1-score   support
           0       0.95      0.98      0.96       100
           1       0.97      0.92      0.94        50
```
- Class 0 (e.g., “No”): 100 actual samples  
- Class 1 (e.g., “Yes”): 50 actual samples

It also shows **accuracy** and **macro/micro averages**.

---

## Quick Summary Table

| Metric | Question it answers |
|--------|----------------------|
| Accuracy | How often is the model correct overall? |
| Confusion matrix | Where exactly are the errors (FP, FN)? |
| Precision | When it says “Yes”, how often is that right? |
| Recall | Of all real “Yes” cases, how many did it find? |
| F1-score | Balance of precision and recall (harmonic mean) |
| Classification report | All of the above (per class) in one view |

---

## Simple Rule of Thumb
- **Accuracy** → balanced data, simple overview.  
- **Confusion matrix** → always look at it — reveals hidden errors.  
- **Precision** → when false alarms are expensive.  
- **Recall** → when missing a positive is expensive.  
- **F1-score** → when classes are imbalanced or you need one balanced metric.  
- **Classification report** → quick full summary of a model's performance.

Great addition — these two are **regression** metrics (for predicting continuous numbers, like price or temperature), while the previous ones were for **classification** (predicting categories, like spam/not spam).

Let me explain them simply.

---

## 1. `r2_score` (R-squared, coefficient of determination)

**What it does:**  
Measures **how well the model's predictions match the actual data** — specifically, how much better your model is than just always predicting the **average**.

**Formula (conceptual):**  
R² = 1 - (SS_res / SS_tot)

- **SS_res** = sum of squared errors (how wrong your model is)
- **SS_tot** = sum of squared differences from the average (how wrong a simple "predict the mean" model would be)

**Interpretation:**
- **R² = 1** → perfect predictions (all errors = 0)
- **R² = 0** → model is as good as just guessing the average
- **R² < 0** → model is worse than guessing the average (very bad)

**Simple way to think:**  
*"What percentage of the variation in the data does my model explain?"*

**Example:**  
Predicting house prices. If R² = 0.85 → your model explains 85% of why prices vary; 15% is unexplained (random noise or missing features).

**When to use:**  
To compare models or understand **goodness of fit**. Doesn't tell you if predictions are biased (always off by a fixed amount).

---

## 2. `mean_squared_error` (MSE)

**What it does:**  
Measures the **average squared difference** between predicted and actual values.

**Formula:**  
MSE = (1/n) × Σ(actual - predicted)²

**Why square?**  
1. Makes **all errors positive** (so positive and negative errors don't cancel out)  
2. **Punishes large errors more** than small ones (because squaring makes big errors much bigger)

**Interpretation:**  
- **MSE = 0** → perfect predictions  
- **Higher MSE** → worse model (more error)  
- Units are **squared** of original units (e.g., if predicting dollars → MSE is in dollars²)

**Example:**  
Actual prices: [100, 200]  
Predicted: [110, 190]  
Errors: [-10, +10]  
Squared: [100, 100]  
MSE = (100+100)/2 = 100 (dollars²)

**When to use:**  
- Want to **heavily penalize large errors** (squaring does this)  
- As a **loss function** for training regression models  
- Compare models (lower MSE = better)

**Cousin metric – RMSE (Root Mean Squared Error):**  
= √MSE → brings units back to original (e.g., dollars instead of dollars²), easier to interpret.

---

## Quick Comparison Table

| Metric | What it measures | Best for | Range | Units |
|--------|----------------|----------|-------|-------|
| **R²** | How much variance the model explains | Comparing models, understanding fit | -∞ to 1 (1 is perfect) | Unitless (percentage-like) |
| **MSE** | Average squared error (penalizes big errors) | Training models, penalizing outliers | 0 to ∞ (0 is perfect) | Squared units |
| **RMSE** (√MSE) | Same as MSE but original units | Human interpretation | 0 to ∞ | Same as prediction units |

---

## Example Together (House Price Prediction)

**Actual prices:** [300k, 400k, 500k]  
**Predicted:** [310k, 390k, 510k]  

- **Errors:** [+10k, -10k, +10k]  
- **Squared errors:** [100M, 100M, 100M] (in k²)  
- **MSE** = (100M + 100M + 100M)/3 = 100M (k²)  
- **RMSE** = √100M = 10k dollars (easy to interpret: "on average, predictions are off by ~$10,000")  

- **Mean of actual:** 400k  
- **SS_tot** = (300-400)² + (400-400)² + (500-400)² = 10k² + 0 + 10k² = 200M  
- **SS_res** = 100M + 100M + 100M = 300M  
- **R²** = 1 - (300M/200M) = 1 - 1.5 = **-0.5** (model is worse than just guessing 400k every time!)

---

## Simple Rule of Thumb

| If you want to... | Use... |
|------------------|--------|
| See how well your model explains the data | R² |
| Compare two regression models | R² (higher = better) |
| Train a model (minimize error) | MSE or RMSE |
| Tell a boss/client about average error | RMSE (because units make sense) |
| Heavily punish rare huge mistakes | MSE (squaring makes big errors huge) |

---

## Final Note — Classification vs Regression Metrics

| Task Type | Metrics |
|-----------|---------|
| **Classification** (yes/no, cat/dog) | accuracy, confusion matrix, precision, recall, f1-score |
| **Regression** (prices, temperatures) | R², MSE, RMSE, MAE (mean absolute error) |

Don't mix them up — R² on a classification problem makes no sense, and accuracy on house prices makes no sense!
